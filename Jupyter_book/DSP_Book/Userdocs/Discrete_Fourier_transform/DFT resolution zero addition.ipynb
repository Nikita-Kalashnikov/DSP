# Разрешающая способность ДПФ, дополнение нулями и дискретизация в частотной области

## Ключевые аспекты

1-Чем больше точек в ДПФ, тем точнее ДПФ аппроксимирует непрерывное преобразование Фурье.

2-Дополнение нулями НЕ увеличивает реальную разрешающую способность она определяется длиной исходного сигнала.

3- Улучшает визуализацию - делает спектр более гладким и удобным для анализ.а

4-Повышает точность оценки частоты - для сигналов с нецелым числом периодов.

5-Позволяет интерполировать сигнал - через операцию в частотной области. Интерполяция это способ оценить промежуточные значения функции на основе уже известных.

6-Упрощает анализ - особенно при работе с алгоритмами, требующими определенной длины сигнала.

7-Дополнение нулями не улучшает способность различать два близко расположенных по частоте сигнала. 

**Зависимость частоты дискретизации от количества отсчетов**

Точные значения частоты разных синусоид зависят как от частоты дискретизации **Fs**, с которой был дискретизирован исходный сигнал, так и от количества отсчетов **N**. Если мы дискретизируем непрерывный сигнал с частотой **Fs** отсчетов в секунду, а затем выполняем **N**-точечное ДПФ дискретизированных данных, основная частота синусоид будет равна **Fanalysis(K) = K*Fs/N**. 

- Fs - Частота отсчетов в секунду.
- K - Анализируемая частота.
- N - Число отсчетов сигнала.

**Пример №1** 

Сигнал дискретизируется с частотой 32 отсчета в секунду. Выполнено 32-точечное дискретное преобразование Фурье сигнала. Сигнал представляет синусоиду частотой 3.7 Гц.
- Fs - 32 отсчета/сек
- K - 3.7 Гц. По причине того, что ДПФ целочисленное, то ближайшая частота будет 4 Гц.
- N - 32
Fanalysis(4) = 4*32/32 = 4 Гц. В чем можно убедиться на графике амплитудного спектра.
Пример продемонстрирован ниже.

**Пример №2** 

Сигнал дискретизируется с частотой 32 отсчета в секунду. Выполнено 128-точечное дискретное преобразование Фурье сигнала. Первые 32 отсчета сигнала - это синусоида частотой 3.7 Гц, а затем 96 отсчетов, равных нулю (32+96=128). Сигнал взят из Примера №1.
- Fs - 32 отсчета/сек
- K - 3.7 Гц. По причине того, что ДПФ целочисленное, то ближайшая частота будет 4 Гц.
- N - 128
- Fanalysis(16) = 16*32/128 = 4 Гц. В чем можно убедиться на графике амплитудного спектра.
Или наоборот:
Поиск 4 Гц на частотной оси: 4*128/32 = 16 - это число на частотной оси будет соответствовать 4 Гц, а для 3.7 Гц: 3.7*128/32 = 14.8 ≈ 15 - это число на частотной оси будет соответствовать примерно 3.7 Гц.
Пример продемонстрирован ниже.

**Пример №3** 

Сигнал дискретизируется с частотой 32 отсчета в секунду. Выполнено 512-точечное дискретное преобразование Фурье сигнала. Первые 32 отсчета сигнала - это синусоида частотой 3.7 Гц, а затем 480 отсчетов, равных нулю (32+480=512). Сигнал взят из Примера №1.
- Fs - 32 отсчета/сек
- K - 3.7 Гц. По причине того, что ДПФ целочисленное, то ближайшая частота будет 4 Гц.
- N - 512
- Fanalysis(59) = 59*32/512 = 3.6875 Гц ≈ 3.7 Гц.

В чем можно убедиться на графике амплитудного спектра.
Или наоборот:
Поиск 3.7 Гц на частотной оси: 3.7*512/32 = 59.2 ≈ 59 - это число на частотной оси будет соответствовать примерно 3.7 Гц.
Пример продемонстрирован ниже.

## Пример №1: 

```python
###-------------------------------------- Пример №1 --------------------------------------###

# Разрешающая способность ДПФ,дополнение нулями и дискретизация в частотной области 
# Чем больше точек в ДПФ, тем точнее ДПФ аппроксимирует непрерывное преобразование Фурье

import numpy as np
import matplotlib.pyplot as plt
from bokeh.layouts import column, gridplot
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
# Активируем вывод графиков в ноутбук
output_notebook()

# ----------------------------
# ПАРАМЕТРЫ СИГНАЛОВ
# ----------------------------

# Общее количество отсчетов для всех сигналов
N = 32  # Количество дискретных точек (отсчетов) для построения сигналов

# Амплитуды сигналов:
W1_A = 1.0  # Амплитуда дополнительной волны 1
W2_A = 1.0  # Амплитуда дополнительной волны 2
W3_A = 1.0  # Амплитуда дополнительной волны 3
W4_A = 1.0  # Амплитуда дополнительной волны 4

# Фазы сигналов в градусах:
W1_P_DEG = 0   # Фаза дополнительной волны 1
W2_P_DEG = 0   # Фаза дополнительной волны 2
W3_P_DEG = 0   # Фаза дополнительной волны 3
W4_P_DEG = 0   # Фаза дополнительной волны 4

# Количество периодов сигнала (K):
W1_K = 3.7  # Дополнительная волна 1
W2_K = 0  # Дополнительная волна 2
W3_K = 0  # Дополнительная волна 3
W4_K = 0  # Дополнительная волна 4

# ----------------------------
# ПОДГОТОВКА ДАННЫХ
# ----------------------------

# Создаем массив отсчетов (от 0 до N-1)
N_i = np.arange(N)  # [0, 1, 2, ..., N-1]

# Конвертация фаз из градусов в радианы (требуется для тригонометрических функций)
W1_P_RAD = W1_P_DEG * np.pi / 180 # Формула преобразования: радианы = градусы * π/180
W2_P_RAD = W2_P_DEG * np.pi / 180
W3_P_RAD = W3_P_DEG * np.pi / 180
W4_P_RAD = W4_P_DEG * np.pi / 180

# ----------------------------
# ГЕНЕРАЦИЯ СИГНАЛОВ
# ----------------------------

# Генерация синусоиды:
# Формула: A * sin(2π * K * n/N + φ)
#   K - количество периодов
#   n - текущий отсчет
#   N - общее количество отсчетов
#   φ - фаза в радианах

# Используем sin вместо cos, но с фазой, отличающейся на π/2
# (cos(x) = sin(x + π/2), но здесь фаза задается параметром)
# Генерация  волн:
wave_1 = W1_A * np.sin(2 * np.pi * W1_K * N_i / N + W1_P_RAD)
wave_2 = W2_A * np.sin(2 * np.pi * W2_K * N_i / N + W2_P_RAD)
wave_3 = W3_A * np.sin(2 * np.pi * W3_K * N_i / N + W3_P_RAD)
wave_4 = W4_A * np.sin(2 * np.pi * W4_K * N_i / N + W4_P_RAD)

# ----------------------------
# СУММИРОВАНИЕ СИГНАЛОВ
# ----------------------------

# Сумма сгенерированных сигналов, в зависимости от примера количество сигналов может быть больше или меньше.
sum_wave = wave_1 + wave_2 + wave_3 + wave_4

# Произвести анализ частотного базиса. С выводом данных о нем.
F_b = int(W1_K)
# DFT (прямое преобразование Фурье)
REX = np.zeros(N)
IMX = np.zeros(N)
PHASE = np.zeros(N)
MAG =  np.zeros(N)
sin_k = np.zeros(N)
cos_k = np.zeros(N)
phase_k = np.zeros(N) 
mag_k =  np.zeros(N)

for K in range(N):
    for I in range(N):
        angle = 2 * np.pi * K * I / N
        REX[K] += sum_wave[I] * np.cos(angle)
        IMX[K] -= sum_wave[I] * np.sin(angle)
        
        if K == F_b:
            cos_k[I] = np.cos(angle)
            sin_k[I] = np.sin(angle)
            # Форматирование значений для выравнивания
            k_str = f"K = {K}"
            i_str = f"I = {I}"
            imx_str = f"IMX[K] = {IMX[K]:.5f}"
            rex_str = f"REX[K] = {REX[K]:.5f}"
            xx_str = f"sum[I] = {sum_wave[I]:.5f}"
            sin_str = f"np.sin(angle) = {np.sin(angle):.5f}"
            cos_str = f"np.cos(angle) = {np.cos(angle):.5f}"
            angle_str = f"angle = {(180 * angle / 3.14) % 360:.2f}°"
            
            # Вывод с фиксированной шириной колонок
            # print(f"{k_str:<8}{i_str:<8}{imx_str:<20}{rex_str:<20}{xx_str:<20}{sin_str:<25}{cos_str:<25}{angle_str:<5}")
    
    # Вычисляем фазу и модуль после завершения внутреннего цикла
    if REX[K] == 0:
        REX[K] = 1e-20  # защита от деления на ноль
    PHASE[K] = np.arctan2(IMX[K], REX[K])  # Используем arctan2 для правильного определения квадранта
    MAG[K] = np.sqrt(REX[K]**2 + IMX[K]**2)
    
    if K == F_b:
        phase_k = PHASE[K]
        mag_k = MAG[K]

# print("Амплитудный и фазовый спектры = " + str(np.sqrt(REX**2 + IMX**2)))

# Амплитудный и фазовый спектры
amplitude_spectrum = np.sqrt(REX**2 + IMX**2)
phase_spectrum = np.arctan2(IMX, REX)

# ----------------------------
# ВИЗУАЛИЗАЦИЯ
# ----------------------------
# Создаем массив для оси X (нулевая линия)
X_axis = np.zeros(N)

# 1. График сигналов и базисных функций
p1 = figure(title="Сигнал и базисные функции", width=800, height=300, x_axis_label='Отсчеты', y_axis_label='Амплитуда')
p1.scatter(N_i, sum_wave, size=7, marker="circle", legend_label="sum_wave", line_color="blue", fill_alpha=0.3)
p1.line(N_i, sum_wave, legend_label="sum_wave", line_color="blue")
p1.scatter(N_i, sin_k, size=7, marker="square", legend_label="sin_k", line_color="green", fill_alpha=0.3)
p1.line(N_i, sin_k, legend_label="sin_k", line_color="green")
p1.line(N_i, X_axis, line_color="black")

# 2. График амплитудного спектра
p2 = figure(title="Амплитудный спектр", width=800, height=300, x_axis_label='Частота', y_axis_label='Амплитуда')
# p2.line(N_i, MAG, legend_label="MAG", line_color="red")
# p2.line(N_i, X_axis, line_color="black")
p2.circle(N_i, MAG, size=8, color="red", alpha=0.7)
p2.segment(x0=N_i, y0=X_axis, x1=N_i, y1=MAG, color="red", alpha=0.6, line_width=2)
# Выделяем точку для K = F_b
# p2.circle([F_b], [mag_k], size=12, color="navy", alpha=1.0, legend_label=f"K={F_b}")

# 3. График фазового спектра
p3 = figure(title="Фазовый спектр", width=800, height=300, x_axis_label='Частота', y_axis_label='Фаза (рад)')
# p3.line(N_i, PHASE, legend_label="PHASE", line_color="purple")
# p3.line(N_i, X_axis, line_color="black")
p3.circle(N_i, PHASE, size=8, color="purple", alpha=0.7)
p3.segment(x0=N_i, y0=X_axis, x1=N_i, y1=PHASE, color="purple", alpha=0.6, line_width=2)
# Выделяем точку для K = F_b
# p3.circle([F_b], [phase_k], size=12, color="navy", alpha=1.0, legend_label=f"K={F_b}")

# 4. График полярного представления для K = F_b
p4 = figure(title=f"Полярное представление для K={F_b}", width=800, height=300,x_axis_label='REX', y_axis_label='IMX')
p4.scatter([REX[F_b]], [IMX[F_b]], size=12, color="navy", alpha=0.7)
p4.segment(x0=[0], y0=[0], x1=[REX[F_b]], y1=[IMX[F_b]], color="navy", alpha=0.6, line_width=2)
# Добавляем annotations
p4.text([REX[F_b]*1.1], [IMX[F_b]*1.1], text=[f"MAG={mag_k:.2f}"], text_color="navy", text_font_size="10pt")
p4.text([REX[F_b]*1.1], [IMX[F_b]*0.9], text=[f"PHASE={phase_k:.2f}"], text_color="navy", text_font_size="10pt")

# 5. НОВЫЙ ГРАФИК: Косинусная и синусная компоненты базисной функции
p5 = figure(title="Базисные функции: cos_k и sin_k", width=800, height=300, x_axis_label='Отсчеты', y_axis_label='Амплитуда')
p5.line(N_i, cos_k, legend_label="cos_k", line_color="orange", line_width=2)
p5.line(N_i, sin_k, legend_label="sin_k", line_color="cyan", line_width=2)
p5.line(N_i, X_axis, line_color="black")
# Добавляем точки для лучшей визуализации
p5.circle(N_i, cos_k, size=5, color="orange", alpha=0.5, legend_label="cos_k")
p5.circle(N_i, sin_k, size=5, color="cyan", alpha=0.5, legend_label="sin_k")

# 6. НОВЫЙ ГРАФИК: Отображение wave_1, wave_2, wave_3, wave_4
p6 = figure(title="Исходные волны: wave_1, wave_2, wave_3, wave_4", width=800, height=300, x_axis_label='Отсчеты', y_axis_label='Амплитуда')
p6.line(N_i, wave_1, legend_label="wave_1", line_color="red", line_width=2)
p6.line(N_i, wave_2, legend_label="wave_2", line_color="green", line_width=2)
p6.line(N_i, wave_3, legend_label="wave_3", line_color="blue", line_width=2)
p6.line(N_i, wave_4, legend_label="wave_4", line_color="purple", line_width=2)
p6.line(N_i, X_axis, line_color="black", line_width=1)

# Добавляем точки для лучшей визуализации
p6.circle(N_i, wave_1, size=4, color="red", alpha=0.5)
p6.circle(N_i, wave_2, size=4, color="green", alpha=0.5)
p6.circle(N_i, wave_3, size=4, color="blue", alpha=0.5)
p6.circle(N_i, wave_4, size=4, color="purple", alpha=0.5)

p6.legend.location = "top_left"
p6.legend.click_policy = "hide"  # Можно кликать по легенде, чтобы скрывать/показывать волны

show(p1)
show(p6)
show(p5)
show(p2)
show(p3)
# show(p4)
# # Собираем все графики в сетку (теперь 3 строки: первые две по 2 графика, третья - один широкий график)
# grid = gridplot([
#     [p1, p2],
#     [p3, p4],
#     [p5, None]  # None означает пустую ячейку
# ])
# show(grid)

# Дополнительная информация в консоли
print(f"\nДля K={F_b}:")
print(f"  MAG[K] = {mag_k}")
print(f"  PHASE[K] = {phase_k} радиан")
print(f"  PHASE[K] = {phase_k * 180/np.pi} градусов")
```

## Пример №2:

```python
###-------------------------------------- Пример №2 --------------------------------------###
# Разрешающая способность ДПФ,дополнение нулями и дискретизация в частотной области 
# Чем больше точек в ДПФ, тем точнее ДПФ аппроксимирует непрерывное преобразование Фурье

import numpy as np
import matplotlib.pyplot as plt
from bokeh.layouts import column, gridplot
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
# Активируем вывод графиков в ноутбук
output_notebook()

# ----------------------------
# ПАРАМЕТРЫ СИГНАЛОВ
# ----------------------------

# Общее количество отсчетов для всех сигналов
N = 128  # Количество дискретных точек (отсчетов) для построения сигналов
N_W = 32 
# Амплитуды сигналов:
W1_A = 1.0  # Амплитуда дополнительной волны 1
W2_A = 1.0  # Амплитуда дополнительной волны 2
W3_A = 1.0  # Амплитуда дополнительной волны 3
W4_A = 1.0  # Амплитуда дополнительной волны 4

# Фазы сигналов в градусах:
W1_P_DEG = 0   # Фаза дополнительной волны 1
W2_P_DEG = 0   # Фаза дополнительной волны 2
W3_P_DEG = 0   # Фаза дополнительной волны 3
W4_P_DEG = 0   # Фаза дополнительной волны 4

# Количество периодов сигнала (K):
# W1_K = 3.7  # Дополнительная волна 1
W1_K = 3.7  # Дополнительная волна 1
W2_K = 0  # Дополнительная волна 2
W3_K = 0  # Дополнительная волна 3
W4_K = 0  # Дополнительная волна 4

# ----------------------------
# ПОДГОТОВКА ДАННЫХ
# ----------------------------

# Создаем массив отсчетов (от 0 до N-1)
N_i = np.arange(N)  # [0, 1, 2, ..., N-1]
N_W_i = np.arange(N_W)  # [0, 1, 2, ..., N_W-1]

# Конвертация фаз из градусов в радианы (требуется для тригонометрических функций)
W1_P_RAD = W1_P_DEG * np.pi / 180 # Формула преобразования: радианы = градусы * π/180
W2_P_RAD = W2_P_DEG * np.pi / 180
W3_P_RAD = W3_P_DEG * np.pi / 180
W4_P_RAD = W4_P_DEG * np.pi / 180

# ----------------------------
# ГЕНЕРАЦИЯ СИГНАЛОВ
# ----------------------------

# Генерация синусоиды:
# Формула: A * sin(2π * K * n/N + φ)
#   K - количество периодов
#   n - текущий отсчет
#   N - общее количество отсчетов
#   φ - фаза в радианах

# Используем sin вместо cos, но с фазой, отличающейся на π/2
# (cos(x) = sin(x + π/2), но здесь фаза задается параметром)
# Генерация  волн:
# Генерация сигналов для первых N_W отсчетов
wave_1_s = W1_A * np.sin(2 * np.pi * W1_K * N_W_i / N_W + W1_P_RAD)
wave_2_s = W2_A * np.sin(2 * np.pi * W2_K * N_W_i / N_W + W2_P_RAD)
wave_3_s = W3_A * np.sin(2 * np.pi * W3_K * N_W_i / N_W + W3_P_RAD)
wave_4_s = W4_A * np.sin(2 * np.pi * W4_K * N_W_i / N_W + W4_P_RAD)

# Создаем полные сигналы длиной N, заполняя нулями оставшуюся часть
wave_1 = np.zeros(N)
wave_2 = np.zeros(N)
wave_3 = np.zeros(N)
wave_4 = np.zeros(N)

# Заполняем первые N_W отсчетов значениями из wave_1_s, wave_2_s и т.д.
wave_1[:N_W] = wave_1_s
wave_2[:N_W] = wave_2_s
wave_3[:N_W] = wave_3_s
wave_4[:N_W] = wave_4_s
# ----------------------------
# СУММИРОВАНИЕ СИГНАЛОВ
# ----------------------------

# Сумма сгенерированных сигналов, в зависимости от примера количество сигналов может быть больше или меньше.
sum_wave = wave_1 + wave_2 + wave_3 + wave_4

# Произвести анализ частотного базиса. С выводом данных о нем.
F_b = int(W1_K)
# DFT (прямое преобразование Фурье)
REX = np.zeros(N)
IMX = np.zeros(N)
PHASE = np.zeros(N)
MAG =  np.zeros(N)
sin_k = np.zeros(N)
cos_k = np.zeros(N)
phase_k = np.zeros(N) 
mag_k =  np.zeros(N)

for K in range(N):
    for I in range(N):
        angle = 2 * np.pi * K * I / N
        REX[K] += sum_wave[I] * np.cos(angle)
        IMX[K] -= sum_wave[I] * np.sin(angle)
        
        if K == F_b:
            cos_k[I] = np.cos(angle)
            sin_k[I] = np.sin(angle)
            # Форматирование значений для выравнивания
            k_str = f"K = {K}"
            i_str = f"I = {I}"
            imx_str = f"IMX[K] = {IMX[K]:.5f}"
            rex_str = f"REX[K] = {REX[K]:.5f}"
            xx_str = f"sum[I] = {sum_wave[I]:.5f}"
            sin_str = f"np.sin(angle) = {np.sin(angle):.5f}"
            cos_str = f"np.cos(angle) = {np.cos(angle):.5f}"
            angle_str = f"angle = {(180 * angle / 3.14) % 360:.2f}°"
            
            # Вывод с фиксированной шириной колонок
            # print(f"{k_str:<8}{i_str:<8}{imx_str:<20}{rex_str:<20}{xx_str:<20}{sin_str:<25}{cos_str:<25}{angle_str:<5}")
    
    # Вычисляем фазу и модуль после завершения внутреннего цикла
    if REX[K] == 0:
        REX[K] = 1e-20  # защита от деления на ноль
    PHASE[K] = np.arctan2(IMX[K], REX[K])  # Используем arctan2 для правильного определения квадранта
    MAG[K] = np.sqrt(REX[K]**2 + IMX[K]**2)
    
    if K == F_b:
        phase_k = PHASE[K]
        mag_k = MAG[K]

# print("Амплитудный и фазовый спектры = " + str(np.sqrt(REX**2 + IMX**2)))

# Амплитудный и фазовый спектры
amplitude_spectrum = np.sqrt(REX**2 + IMX**2)
phase_spectrum = np.arctan2(IMX, REX)

# ----------------------------
# ВИЗУАЛИЗАЦИЯ
# ----------------------------
# Создаем массив для оси X (нулевая линия)
X_axis = np.zeros(N)

# 1. График сигналов и базисных функций
p1 = figure(title="Сигнал и базисные функции", width=800, height=300, x_axis_label='Отсчеты', y_axis_label='Амплитуда')
p1.scatter(N_i, sum_wave, size=7, marker="circle", legend_label="sum_wave", line_color="blue", fill_alpha=0.3)
p1.line(N_i, sum_wave, legend_label="sum_wave", line_color="blue")
p1.scatter(N_i, sin_k, size=7, marker="square", legend_label="sin_k", line_color="green", fill_alpha=0.3)
p1.line(N_i, sin_k, legend_label="sin_k", line_color="green")
p1.line(N_i, X_axis, line_color="black")

# 2. График амплитудного спектра
p2 = figure(title="Амплитудный спектр", width=800, height=300, x_axis_label='Частота', y_axis_label='Амплитуда')
# p2.line(N_i, MAG, legend_label="MAG", line_color="red")
# p2.line(N_i, X_axis, line_color="black")
p2.circle(N_i, MAG, size=8, color="red", alpha=0.7)
p2.segment(x0=N_i, y0=X_axis, x1=N_i, y1=MAG, color="red", alpha=0.6, line_width=2)
# Выделяем точку для K = F_b
# p2.circle([F_b], [mag_k], size=12, color="navy", alpha=1.0, legend_label=f"K={F_b}")

# 3. График фазового спектра
p3 = figure(title="Фазовый спектр", width=800, height=300, x_axis_label='Частота', y_axis_label='Фаза (рад)')
# p3.line(N_i, PHASE, legend_label="PHASE", line_color="purple")
# p3.line(N_i, X_axis, line_color="black")
p3.circle(N_i, PHASE, size=8, color="purple", alpha=0.7)
p3.segment(x0=N_i, y0=X_axis, x1=N_i, y1=PHASE, color="purple", alpha=0.6, line_width=2)
# Выделяем точку для K = F_b
# p3.circle([F_b], [phase_k], size=12, color="navy", alpha=1.0, legend_label=f"K={F_b}")

# 4. График полярного представления для K = F_b
p4 = figure(title=f"Полярное представление для K={F_b}", width=800, height=300,x_axis_label='REX', y_axis_label='IMX')
p4.scatter([REX[F_b]], [IMX[F_b]], size=12, color="navy", alpha=0.7)
p4.segment(x0=[0], y0=[0], x1=[REX[F_b]], y1=[IMX[F_b]], color="navy", alpha=0.6, line_width=2)
# Добавляем annotations
p4.text([REX[F_b]*1.1], [IMX[F_b]*1.1], text=[f"MAG={mag_k:.2f}"], text_color="navy", text_font_size="10pt")
p4.text([REX[F_b]*1.1], [IMX[F_b]*0.9], text=[f"PHASE={phase_k:.2f}"], text_color="navy", text_font_size="10pt")

# 5. НОВЫЙ ГРАФИК: Косинусная и синусная компоненты базисной функции
p5 = figure(title="Базисные функции: cos_k и sin_k", width=800, height=300, x_axis_label='Отсчеты', y_axis_label='Амплитуда')
p5.line(N_i, cos_k, legend_label="cos_k", line_color="orange", line_width=2)
p5.line(N_i, sin_k, legend_label="sin_k", line_color="cyan", line_width=2)
p5.line(N_i, X_axis, line_color="black")
# Добавляем точки для лучшей визуализации
p5.circle(N_i, cos_k, size=5, color="orange", alpha=0.5, legend_label="cos_k")
p5.circle(N_i, sin_k, size=5, color="cyan", alpha=0.5, legend_label="sin_k")

# 6. НОВЫЙ ГРАФИК: Отображение wave_1, wave_2, wave_3, wave_4
p6 = figure(title="Исходные волны: wave_1, wave_2, wave_3, wave_4", width=800, height=300, x_axis_label='Отсчеты', y_axis_label='Амплитуда')
p6.line(N_i, wave_1, legend_label="wave_1", line_color="red", line_width=2)
p6.line(N_i, wave_2, legend_label="wave_2", line_color="green", line_width=2)
p6.line(N_i, wave_3, legend_label="wave_3", line_color="blue", line_width=2)
p6.line(N_i, wave_4, legend_label="wave_4", line_color="purple", line_width=2)
p6.line(N_i, X_axis, line_color="black", line_width=1)

# Добавляем точки для лучшей визуализации
p6.circle(N_i, wave_1, size=4, color="red", alpha=0.5)
p6.circle(N_i, wave_2, size=4, color="green", alpha=0.5)
p6.circle(N_i, wave_3, size=4, color="blue", alpha=0.5)
p6.circle(N_i, wave_4, size=4, color="purple", alpha=0.5)

p6.legend.location = "top_left"
p6.legend.click_policy = "hide"  # Можно кликать по легенде, чтобы скрывать/показывать волны

show(p1)
show(p6)
show(p5)
show(p2)
show(p3)
# show(p4)
# # Собираем все графики в сетку (теперь 3 строки: первые две по 2 графика, третья - один широкий график)
# grid = gridplot([
#     [p1, p2],
#     [p3, p4],
#     [p5, None]  # None означает пустую ячейку
# ])
# show(grid)

# Дополнительная информация в консоли
print(f"\nДля K={F_b}:")
print(f"  MAG[K] = {mag_k}")
print(f"  PHASE[K] = {phase_k} радиан")
print(f"  PHASE[K] = {phase_k * 180/np.pi} градусов")
```

```python
###-------------------------------------- Пример №3 --------------------------------------###
# Разрешающая способность ДПФ,дополнение нулями и дискретизация в частотной области 
# Чем больше точек в ДПФ, тем точнее ДПФ аппроксимирует непрерывное преобразование Фурье

import numpy as np
import matplotlib.pyplot as plt
from bokeh.layouts import column, gridplot
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
# Активируем вывод графиков в ноутбук
output_notebook()

# ----------------------------
# ПАРАМЕТРЫ СИГНАЛОВ
# ----------------------------

# Общее количество отсчетов для всех сигналов
N = 512  # Количество дискретных точек (отсчетов) для построения сигналов
N_W = 32 
# Амплитуды сигналов:
W1_A = 1.0  # Амплитуда дополнительной волны 1
W2_A = 1.0  # Амплитуда дополнительной волны 2
W3_A = 1.0  # Амплитуда дополнительной волны 3
W4_A = 1.0  # Амплитуда дополнительной волны 4

# Фазы сигналов в градусах:
W1_P_DEG = 0   # Фаза дополнительной волны 1
W2_P_DEG = 0   # Фаза дополнительной волны 2
W3_P_DEG = 0   # Фаза дополнительной волны 3
W4_P_DEG = 0   # Фаза дополнительной волны 4

# Количество периодов сигнала (K):
# W1_K = 3.7  # Дополнительная волна 1
W1_K = 3.7  # Дополнительная волна 1
W2_K = 0  # Дополнительная волна 2
W3_K = 0  # Дополнительная волна 3
W4_K = 0  # Дополнительная волна 4

# ----------------------------
# ПОДГОТОВКА ДАННЫХ
# ----------------------------

# Создаем массив отсчетов (от 0 до N-1)
N_i = np.arange(N)  # [0, 1, 2, ..., N-1]
N_W_i = np.arange(N_W)  # [0, 1, 2, ..., N_W-1]

# Конвертация фаз из градусов в радианы (требуется для тригонометрических функций)
W1_P_RAD = W1_P_DEG * np.pi / 180 # Формула преобразования: радианы = градусы * π/180
W2_P_RAD = W2_P_DEG * np.pi / 180
W3_P_RAD = W3_P_DEG * np.pi / 180
W4_P_RAD = W4_P_DEG * np.pi / 180

# ----------------------------
# ГЕНЕРАЦИЯ СИГНАЛОВ
# ----------------------------

# Генерация синусоиды:
# Формула: A * sin(2π * K * n/N + φ)
#   K - количество периодов
#   n - текущий отсчет
#   N - общее количество отсчетов
#   φ - фаза в радианах

# Используем sin вместо cos, но с фазой, отличающейся на π/2
# (cos(x) = sin(x + π/2), но здесь фаза задается параметром)
# Генерация  волн:
# Генерация сигналов для первых N_W отсчетов
wave_1_s = W1_A * np.sin(2 * np.pi * W1_K * N_W_i / N_W + W1_P_RAD)
wave_2_s = W2_A * np.sin(2 * np.pi * W2_K * N_W_i / N_W + W2_P_RAD)
wave_3_s = W3_A * np.sin(2 * np.pi * W3_K * N_W_i / N_W + W3_P_RAD)
wave_4_s = W4_A * np.sin(2 * np.pi * W4_K * N_W_i / N_W + W4_P_RAD)

# Создаем полные сигналы длиной N, заполняя нулями оставшуюся часть
wave_1 = np.zeros(N)
wave_2 = np.zeros(N)
wave_3 = np.zeros(N)
wave_4 = np.zeros(N)

# Заполняем первые N_W отсчетов значениями из wave_1_s, wave_2_s и т.д.
wave_1[:N_W] = wave_1_s
wave_2[:N_W] = wave_2_s
wave_3[:N_W] = wave_3_s
wave_4[:N_W] = wave_4_s
# ----------------------------
# СУММИРОВАНИЕ СИГНАЛОВ
# ----------------------------

# Сумма сгенерированных сигналов, в зависимости от примера количество сигналов может быть больше или меньше.
sum_wave = wave_1 + wave_2 + wave_3 + wave_4

# Произвести анализ частотного базиса. С выводом данных о нем.
F_b = int(W1_K)
# DFT (прямое преобразование Фурье)
REX = np.zeros(N)
IMX = np.zeros(N)
PHASE = np.zeros(N)
MAG =  np.zeros(N)
sin_k = np.zeros(N)
cos_k = np.zeros(N)
phase_k = np.zeros(N) 
mag_k =  np.zeros(N)

for K in range(N):
    for I in range(N):
        angle = 2 * np.pi * K * I / N
        REX[K] += sum_wave[I] * np.cos(angle)
        IMX[K] -= sum_wave[I] * np.sin(angle)
        
        if K == F_b:
            cos_k[I] = np.cos(angle)
            sin_k[I] = np.sin(angle)
            # Форматирование значений для выравнивания
            k_str = f"K = {K}"
            i_str = f"I = {I}"
            imx_str = f"IMX[K] = {IMX[K]:.5f}"
            rex_str = f"REX[K] = {REX[K]:.5f}"
            xx_str = f"sum[I] = {sum_wave[I]:.5f}"
            sin_str = f"np.sin(angle) = {np.sin(angle):.5f}"
            cos_str = f"np.cos(angle) = {np.cos(angle):.5f}"
            angle_str = f"angle = {(180 * angle / 3.14) % 360:.2f}°"
            
            # Вывод с фиксированной шириной колонок
            # print(f"{k_str:<8}{i_str:<8}{imx_str:<20}{rex_str:<20}{xx_str:<20}{sin_str:<25}{cos_str:<25}{angle_str:<5}")
    
    # Вычисляем фазу и модуль после завершения внутреннего цикла
    if REX[K] == 0:
        REX[K] = 1e-20  # защита от деления на ноль
    PHASE[K] = np.arctan2(IMX[K], REX[K])  # Используем arctan2 для правильного определения квадранта
    MAG[K] = np.sqrt(REX[K]**2 + IMX[K]**2)
    
    if K == F_b:
        phase_k = PHASE[K]
        mag_k = MAG[K]

# print("Амплитудный и фазовый спектры = " + str(np.sqrt(REX**2 + IMX**2)))

# Амплитудный и фазовый спектры
amplitude_spectrum = np.sqrt(REX**2 + IMX**2)
phase_spectrum = np.arctan2(IMX, REX)

# ----------------------------
# ВИЗУАЛИЗАЦИЯ
# ----------------------------
# Создаем массив для оси X (нулевая линия)
X_axis = np.zeros(N)

# 1. График сигналов и базисных функций
p1 = figure(title="Сигнал и базисные функции", width=800, height=300, x_axis_label='Отсчеты', y_axis_label='Амплитуда')
p1.scatter(N_i, sum_wave, size=7, marker="circle", legend_label="sum_wave", line_color="blue", fill_alpha=0.3)
p1.line(N_i, sum_wave, legend_label="sum_wave", line_color="blue")
p1.scatter(N_i, sin_k, size=7, marker="square", legend_label="sin_k", line_color="green", fill_alpha=0.3)
p1.line(N_i, sin_k, legend_label="sin_k", line_color="green")
p1.line(N_i, X_axis, line_color="black")

# 2. График амплитудного спектра
p2 = figure(title="Амплитудный спектр", width=800, height=300, x_axis_label='Частота', y_axis_label='Амплитуда')
# p2.line(N_i, MAG, legend_label="MAG", line_color="red")
# p2.line(N_i, X_axis, line_color="black")
p2.circle(N_i, MAG, size=8, color="red", alpha=0.7)
p2.segment(x0=N_i, y0=X_axis, x1=N_i, y1=MAG, color="red", alpha=0.6, line_width=2)
# Выделяем точку для K = F_b
# p2.circle([F_b], [mag_k], size=12, color="navy", alpha=1.0, legend_label=f"K={F_b}")

# 3. График фазового спектра
p3 = figure(title="Фазовый спектр", width=800, height=300, x_axis_label='Частота', y_axis_label='Фаза (рад)')
# p3.line(N_i, PHASE, legend_label="PHASE", line_color="purple")
# p3.line(N_i, X_axis, line_color="black")
p3.circle(N_i, PHASE, size=8, color="purple", alpha=0.7)
p3.segment(x0=N_i, y0=X_axis, x1=N_i, y1=PHASE, color="purple", alpha=0.6, line_width=2)
# Выделяем точку для K = F_b
# p3.circle([F_b], [phase_k], size=12, color="navy", alpha=1.0, legend_label=f"K={F_b}")

# 4. График полярного представления для K = F_b
p4 = figure(title=f"Полярное представление для K={F_b}", width=800, height=300,x_axis_label='REX', y_axis_label='IMX')
p4.scatter([REX[F_b]], [IMX[F_b]], size=12, color="navy", alpha=0.7)
p4.segment(x0=[0], y0=[0], x1=[REX[F_b]], y1=[IMX[F_b]], color="navy", alpha=0.6, line_width=2)
# Добавляем annotations
p4.text([REX[F_b]*1.1], [IMX[F_b]*1.1], text=[f"MAG={mag_k:.2f}"], text_color="navy", text_font_size="10pt")
p4.text([REX[F_b]*1.1], [IMX[F_b]*0.9], text=[f"PHASE={phase_k:.2f}"], text_color="navy", text_font_size="10pt")

# 5. НОВЫЙ ГРАФИК: Косинусная и синусная компоненты базисной функции
p5 = figure(title="Базисные функции: cos_k и sin_k", width=800, height=300, x_axis_label='Отсчеты', y_axis_label='Амплитуда')
p5.line(N_i, cos_k, legend_label="cos_k", line_color="orange", line_width=2)
p5.line(N_i, sin_k, legend_label="sin_k", line_color="cyan", line_width=2)
p5.line(N_i, X_axis, line_color="black")
# Добавляем точки для лучшей визуализации
p5.circle(N_i, cos_k, size=5, color="orange", alpha=0.5, legend_label="cos_k")
p5.circle(N_i, sin_k, size=5, color="cyan", alpha=0.5, legend_label="sin_k")

# 6. НОВЫЙ ГРАФИК: Отображение wave_1, wave_2, wave_3, wave_4
p6 = figure(title="Исходные волны: wave_1, wave_2, wave_3, wave_4", width=800, height=300, x_axis_label='Отсчеты', y_axis_label='Амплитуда')
p6.line(N_i, wave_1, legend_label="wave_1", line_color="red", line_width=2)
p6.line(N_i, wave_2, legend_label="wave_2", line_color="green", line_width=2)
p6.line(N_i, wave_3, legend_label="wave_3", line_color="blue", line_width=2)
p6.line(N_i, wave_4, legend_label="wave_4", line_color="purple", line_width=2)
p6.line(N_i, X_axis, line_color="black", line_width=1)

# Добавляем точки для лучшей визуализации
p6.circle(N_i, wave_1, size=4, color="red", alpha=0.5)
p6.circle(N_i, wave_2, size=4, color="green", alpha=0.5)
p6.circle(N_i, wave_3, size=4, color="blue", alpha=0.5)
p6.circle(N_i, wave_4, size=4, color="purple", alpha=0.5)

p6.legend.location = "top_left"
p6.legend.click_policy = "hide"  # Можно кликать по легенде, чтобы скрывать/показывать волны

show(p1)
show(p6)
show(p5)
show(p2)
show(p3)
# show(p4)
# # Собираем все графики в сетку (теперь 3 строки: первые две по 2 графика, третья - один широкий график)
# grid = gridplot([
#     [p1, p2],
#     [p3, p4],
#     [p5, None]  # None означает пустую ячейку
# ])
# show(grid)

# Дополнительная информация в консоли
print(f"\nДля K={F_b}:")
print(f"  MAG[K] = {mag_k}")
print(f"  PHASE[K] = {phase_k} радиан")
print(f"  PHASE[K] = {phase_k * 180/np.pi} градусов")
```

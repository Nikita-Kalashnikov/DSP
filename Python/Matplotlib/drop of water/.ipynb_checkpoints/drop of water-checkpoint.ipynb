```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib.patches import Circle
import matplotlib.colors as mcolors

# Настройка фигуры
fig, ax = plt.subplots(figsize=(8, 6), facecolor='black')
ax.set_facecolor('black')
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1, 1)
ax.set_aspect('equal')
ax.axis('off')

# Создание воды (синий фон)
water = plt.Rectangle((-2, -1), 4, 1, color='#0a0a2a', alpha=1)
ax.add_patch(water)

# Коллекции для элементов анимации
ripples = []
wave_circles = []
drop_circles = []

# Цвета для волн
colors = ['#1f77b4', '#1e90ff', '#00bfff', '#87cefa', '#b0e0e6']

# Функция инициализации
def init():
    return []

# Функция обновления кадра
def update(frame):
    global ripples, wave_circles, drop_circles
    
    # Очищаем предыдущие волны
    for circle in wave_circles:
        circle.remove()
    wave_circles = []
    
    # Создаем новую каплю каждые 100 кадров
    if frame % 100 == 0:
        x = np.random.uniform(-1.0, 1.0)
        y = 0.8
        size = np.random.uniform(0.05, 0.1)
        ripples.append({
            'x': x, 
            'y': y,
            'size': size,
            'start_frame': frame,
            'waves': []
        })
    
    # Обновляем существующие капли и волны
    for ripple in ripples[:]:
        # Падение капли
        fall_speed = 0.03
        if ripple['y'] > -0.8:
            ripple['y'] -= fall_speed
            
            # Обновляем позицию капли
            drop = Circle((ripple['x'], ripple['y']), ripple['size'], 
                         color='white', alpha=0.8)
            ax.add_patch(drop)
            drop_circles.append(drop)
        
        # Создаем волну при касании воды
        if abs(ripple['y'] + 0.8) < 0.02 and frame - ripple['start_frame'] < 10:
            ripple['waves'].append({
                'radius': 0.01,
                'thickness': 0.02,
                'alpha': 1.0,
                'color': np.random.choice(colors)
            })
        
        # Обновляем волны
        for wave in ripple['waves'][:]:
            # Увеличиваем радиус и уменьшаем прозрачность
            wave['radius'] += 0.03
            wave['thickness'] += 0.002
            wave['alpha'] *= 0.93
            
            # Создаем волну
            wave_circle = Circle((ripple['x'], -0.8), wave['radius'],
                                fill=False, 
                                linewidth=wave['thickness'],
                                edgecolor=wave['color'],
                                alpha=wave['alpha'])
            ax.add_patch(wave_circle)
            wave_circles.append(wave_circle)
            
            # Удаляем старые волны
            if wave['alpha'] < 0.05:
                ripple['waves'].remove(wave)
    
    # Удаляем старые капли
    if len(ripples) > 5:
        ripples.pop(0)
    
    return wave_circles + drop_circles

# Создание анимации
ani = FuncAnimation(fig, update, frames=500, init_func=init,
                    blit=True, interval=20)

plt.tight_layout()
plt.show()

# Сохранение в GIF
ani.save('water_ripples.gif', writer='pillow', fps=30, dpi=80)
```

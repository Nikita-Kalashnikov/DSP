```python
'''Рендеринг кругов'''
# from bokeh.io import output_notebook
# # Активируем вывод графиков в jupytr notebook
# output_notebook()


# Создание простой линейной диараммы 
from bokeh.plotting import figure, show

# Используйте scatter()функцию вместо line()для отрисовки кругов. Круги — это тип маркера по умолчанию для функции рассеивания.
# p.scatter(x, y3, legend_label="Objects", color="yellow", size=12)

# подготовьте данные
x = [1, 2, 3, 4, 5]
y1 = [6, 7, 2, 4, 5]
y2 = [2, 3, 4, 5, 6]
y3 = [4, 5, 5, 7, 2]

# создайте новый график с заголовком и метками осей
p = figure(title="Пример нескольких глифов", x_axis_label="x", y_axis_label="y")

# добавить несколько средств визуализации
p.line(x, y1, legend_label="Температура.", color="blue", line_width=2)
p.vbar(x=x, top=y2, legend_label="курс", width=0.5, bottom=0, color="red")
p.scatter(x, y3, legend_label="Объекты", color="yellow", size=12)

# показать результаты
show(p)
```

```python
'''Настройка символов'''
# from bokeh.io import output_notebook
# # Активируем вывод графиков в jupytr notebook
# output_notebook()

# цвета https://www.w3.org/TR/css-color-4/#named-colors
# Создание простой линейной диараммы 
from bokeh.plotting import figure, show

# Функция scatter(), например, позволяет определять такие аспекты, как цвет или диаметр кругов:
# fill_color: цвет заливки кругов
# fill_alpha: прозрачность цвета заливки (любое значение между 0 и 1)
# line_color: цвет заливки контуров кругов
# size: размер кругов (в единицах экрана )
# legend_label: запись легенды для кругов
# Функция scatter()также поддерживает другие типы маркеров:
# marker: название выбранной формы, например, «круг», «квадрат», «y». MarkerTypeПолный список см. здесь.
# Обратите внимание, что в предыдущих примерах вы использовали colorсвойство для определения цвета объекта. color — 
# это псевдоним, который автоматически устанавливает все цветовые свойства объекта в один и тот же цвет. Например, 
# передача свойства "yellow"круга colorэквивалентна установке fill_colorи line_colorдля жёлтого цвета по отдельности.
# В эффекте «Боке» цвета можно задать несколькими способами. Например:
# Используйте один из именованных цветов CSS (например, "firebrick")
# Используйте шестнадцатеричные значения, начинающиеся с символа #(например "#00ff00", )
# Используйте кортеж из 3 цветов для RGB (например,(100, 100, 255)
# Используйте 4-кортеж для цветов RGBA (например )(100, 100, 255, 0.5)
# Создайте круги с надписью «Объекты» и сделайте их слегка прозрачными с красной заливкой и синими контурами:

from bokeh.plotting import figure, show

# prepare some data
x = [1, 2, 3, 4, 5]
y = [4, 5, 5, 7, 2]

# create a new plot with a title and axis labels
p = figure(title="Glyphs properties example", x_axis_label="x", y_axis_label="y")

# цвета https://www.w3.org/TR/css-color-4/#named-colors
# add circle renderer with additional arguments
p.scatter(
    x,
    y,
    marker="circle",

    # size: размер кругов (в единицах экрана )
    size=40,
    legend_label="Objects",
    # fill_color: цвет заливки кругов - цвета https://www.w3.org/TR/css-color-4/#named-colors
    # fill_color=(10, 10, 20, 0.9),
    fill_color = "#00ff00", # зеленый
    # fill_color = "#8a2be2", # фиолетовый
    
    # fill_alpha: прозрачность цвета заливки (любое значение между 0 и 1)
    # fill_alpha=0.5,
    fill_alpha=1.0,

    # line_color="blue",
    line_color = "#0000cd",
)

# show the results
show(p)
```
